### Starter 2：`def ex1_temp_convert(c):`
（教师版：下方为 TODO 完成态，可运行；学生版此处留空）

````
def ex1_temp_convert(c):
    """题1(⭐): 摄氏→(华氏, 开尔文)  F=C*9/5+32  K=C+273.15"""
    # === 教师版完成态(学生版此处留空待填) ===
    f = c * 9 / 5 + 32          # 9/5 必须写成除法: 写成 9//5 会得 1
    k = c + 273.15
    return f, k
````

### 💡 分层提示（先自己试5分钟再看）
**思路提示**
> - 题1：先写公式再拆成两个变量，返回元组；`9/5` 是真除，别写成 `9//5`
> - 题2：`d = b*b - 4*a*c`；`d < 0` 直接 `return None`，否则 `math.sqrt(d)` 取正负两个根
> - 题3：增 = `devices[k] = {...}`；改 = `devices["DEV-02"]["status"] = "运行"`；删 = `devices.pop("DEV-01")`
> - 题4：子类 `__init__` 先 `super().__init__(dev_id, name)` 继承父类字段，再补 `self.scan_ms`，并覆盖 `start()`
> - 题5：`try` 里 `open(..., encoding="utf-8")`，`except FileNotFoundError` / `except UnicodeDecodeError` 各返回约定字符串

### Starter 3：`def ex2_quadratic(a, b, c):`
（教师版：下方为 TODO 完成态，可运行；学生版此处留空）

````
def ex2_quadratic(a, b, c):
    """题2(⭐⭐): 一元二次方程, 返回根的元组; 无实根返回None"""
    # === 教师版已完成(含判别式分支) ===
    import math
    d = b * b - 4 * a * c               # 判别式
    if d < 0:
        return None                     # 无实根
    r = math.sqrt(d)
    return ((-b + r) / (2 * a), (-b - r) / (2 * a))
````

### Starter 4：`def ex3_device_manager():`
（教师版：下方为 TODO 完成态，可运行；学生版此处留空）

````
def ex3_device_manager():
    """题3(⭐⭐): 设备增删改查(字典), 完成四个操作并返回最终字典"""
    devices = {}                     # id -> {"name":..., "status":...}
    # 1) 增加 DEV-01 空压机 运行
    # 2) 增加 DEV-02 冷却泵 停止
    # 3) 修改 DEV-02 状态为运行
    # 4) 删除 DEV-01
    # === 教师版已完成 ===
    devices["DEV-01"] = {"name": "空压机", "status": "运行"}
    devices["DEV-02"] = {"name": "冷却泵", "status": "停止"}
    devices["DEV-02"]["status"] = "运行"      # 改: 只动嵌套字典的 value
    devices.pop("DEV-01")                      # 删: 键可能不存在时用 pop(k, None) 更稳
    return devices
````

### Starter 5：`class Device:                        # 题4用`
（教师版：下方为 TODO 完成态，可运行；学生版此处留空）

````
class Device:                        # 题4用
    def __init__(self, dev_id, name):
        self.dev_id, self.name = dev_id, name
        self.status = "停止"
````

### Starter 6：`def ex4_plc_class():`
（教师版：下方为 TODO 完成态，可运行；学生版此处留空）

````
def ex4_plc_class():
    """题4(⭐⭐⭐): Device基类 + PLC子类(新增scan_ms, 覆盖start) → 返回PLC实例"""
    class PLC(Device):
        # === 教师版已完成 ===
        def __init__(self, dev_id, name, scan_ms):
            super().__init__(dev_id, name)      # 复用基类: dev_id/name/status
            self.scan_ms = scan_ms              # 子类新增字段

        def start(self):                        # 覆盖基类 start
            self.status = "运行"
            return f"{self.name}启动, 扫描{self.scan_ms}ms"

    return PLC("DEV-01", "S7-1200", 8)
````

### Starter 7：`def ex5_safe_reader(path):`
（教师版：下方为 TODO 完成态，可运行；学生版此处留空）

````
def ex5_safe_reader(path):
    """题5(⭐⭐): 读文件返回内容; 不存在返回"FILE_NOT_FOUND"; 编码错返回"BAD_ENCODING"
    必须用try/except, 禁止os.path.exists预判"""
    # === 教师版已完成 ===
    try:
        with open(path, encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        return "FILE_NOT_FOUND"
    except UnicodeDecodeError:
        return "BAD_ENCODING"
````

### B层：命名规范 + PEP8 打分（工单达标线 ≥85%）

> 本机 `pp03` 环境**没有装 flake8 / pylint / black**（只有 pytest），所以这里用标准库
> `re` 复现 6 条最常考规则，口径完全公开、可自查：
> 行宽≤79 · 缩进4空格无Tab · 逗号后有空格 · 函数/变量 snake_case · 类名 PascalCase ·
> 禁裸 `except:` / 禁 `== None`。装了 flake8 的机器可用 `python -m flake8 --max-line-length=79` 交叉复核。


````
# === B层验收：教学口径 PEP8/命名规范检查（工单达标线 85%）===
import json
import os
import re

RULES = ("行宽≤79", "缩进4空格且无Tab", "逗号后要有空格",
         "函数/变量 snake_case", "类名 PascalCase", "禁裸except/禁== None")


def cell_source(starts):
    """从 notebook 自己的 JSON 里取出以 starts 开头的那个代码格原文。
    不用 inspect.getsource：类对象在 IPython 内核里会去 __main__.__file__（=内核启动脚本）
    找定义，直接抛 "could not find class definition"。读格子文本才是被测的那份代码。
    同名出现多格时取最短的一格（题4基类在答案附录里又写了一次，那格还捎带 p11()）。"""
    if not (NB_PATH and os.path.exists(NB_PATH)):
        return None
    hits = []
    for c in json.load(open(NB_PATH, encoding="utf-8"))["cells"]:
        if c["cell_type"] != "code":
            continue
        s = "".join(c["source"])
        body = "\n".join(x for x in s.splitlines()
                         if x.strip() and not x.lstrip().startswith("#"))
        if body.startswith(starts):
            hits.append(s)
    return min(hits, key=len) if hits else None


def strip_literals(code):
    """把字符串/文档串占位成 ''，避免中文串里的逗号、冒号造成误报。"""
    return re.sub(r'"""[\s\S]*?"""|"[^"]*"|\'[^\']*\'', "''", code)


def lint(src, label):
    bad, checked = [], 0
    for n, line in enumerate(src.splitlines(), 1):
        checked += 1
        bare = strip_literals(line.split("#")[0])
        if len(line) > 79:
            bad.append("L%d %s: 行宽%d>79" % (n, label, len(line)))
        ind = line[:len(line) - len(line.lstrip())]
        if "\t" in ind:
            bad.append("L%d %s: 缩进含 Tab" % (n, label))
        elif len(ind) % 4:
            bad.append("L%d %s: 缩进%d格不是4的倍数" % (n, label, len(ind)))
        if re.search(r",(?![ ]|$)", bare):
            bad.append("L%d %s: 逗号后缺空格" % (n, label))
        if re.search(r"\bexcept\s*:", bare):
            bad.append("L%d %s: 裸 except 吞掉一切异常" % (n, label))
        if re.search(r"==\s*None\b", bare):
            bad.append("L%d %s: 应写 is None" % (n, label))
        for fn in re.findall(r"\bdef\s+([A-Za-z_]\w*)", bare):
            if re.fullmatch(r"__\w+__", fn):        # PEP8：双下划线前后缀是魔术方法，合法
                continue
            if not re.fullmatch(r"_?[a-z][a-z0-9_]*", fn):
                bad.append("L%d %s: 函数名 %s 不是 snake_case" % (n, label, fn))
        for cn in re.findall(r"\bclass\s+([A-Za-z_]\w*)", bare):
            if not re.fullmatch(r"[A-Z][A-Za-z0-9]*", cn):
                bad.append("L%d %s: 类名 %s 不是 PascalCase" % (n, label, cn))
        for tgt in re.findall(r"^([a-zA-Z_]\w*)\s*=[^=]", bare.strip()):
            if not re.fullmatch(r"_?[a-z][a-z0-9_]*|[A-Z][A-Z0-9_]*", tgt):
                bad.append("L%d %s: 变量 %s 命名不规范" % (n, label, tgt))
    return bad, checked, 100.0 * (checked - len(bad)) / max(checked, 1)


targets = [("def ex1_temp_convert", "ex1"), ("def ex2_quadratic", "ex2"),
           ("def ex3_device_manager", "ex3"), ("def ex4_plc_class", "ex4"),
           ("def ex5_safe_reader", "ex5"), ("class Device", "Device")]
print("被测代码来源：notebook 自身格子原文（%s）" % os.path.basename(NB_PATH or "无"))
print("检查规则（教学口径 6 条）:", " · ".join(RULES))
total_bad = total_lines = 0
for starts, label in targets:
    src = cell_source(starts)
    if src is None:
        print("  ✗ %s 没找到格子 %s —— B层证据不成立" % (label, starts))
        continue
    bad, lines, score = lint(src, label)
    total_bad += len(bad)
    total_lines += lines
    print("  %-18s %2d 行  违规 %d  得分 %5.1f%%  %s"
          % (label, lines, len(bad), score, "✔" if score >= 85 else "✗"))
    for b in bad[:4]:
        print("        -", b)
assert total_lines, "一格都没取到，B层闸门不判过"
overall = 100.0 * (total_lines - total_bad) / total_lines
print("PEP8/命名规范（教学口径）总评: %d 行 / %d 违规 = %.1f%%  达标线 85%% → %s"
      % (total_lines, total_bad, overall, "B层达标 ✔" if overall >= 85 else "未达标 ✗"))
assert overall >= 85.0, "B层 PEP8≥85% 闸门未过，先按上面的明细逐条改"
````

### A层：每题 3 条边界断言 + pytest 全绿

工单A层要求「每题附3边界断言、pytest全绿、挑1条向全班讲解」。
下面把**教师版格子里的真代码**用 `inspect.getsource` 抽出来写成临时模块再跑 pytest——
所以测的就是你刚看到的那份实现，不存在"测试和答案各写一套"。


````
# === A层验收：五题各3条边界断言，落到真 pytest（工单达标：pytest全绿）===
import json
import os as _os
import re
import subprocess
import sys
import tempfile


def cell_source(starts):
    """取 notebook 自身格子原文（同 B 层口径：类不能用 inspect.getsource）。"""
    hits = []
    for c in json.load(open(NB_PATH, encoding="utf-8"))["cells"]:
        if c["cell_type"] != "code":
            continue
        s = "".join(c["source"])
        body = "\n".join(x for x in s.splitlines()
                         if x.strip() and not x.lstrip().startswith("#"))
        if body.startswith(starts):
            hits.append(s)
    return min(hits, key=len)


parts = [cell_source(p) for p in ("class Device", "def ex1_temp_convert", "def ex2_quadratic",
                                  "def ex3_device_manager", "def ex4_plc_class",
                                  "def ex5_safe_reader")]
assert all(parts), "有题的格子没取到，A层测试不成立"
_mod_src = "\n".join(["# -*- coding: utf-8 -*-",
                      "# 本模块由 A 层格子从 notebook 自身逐格原文导出，测的就是上方实现",
                      "import math", ""] + parts)

_test_src = r"""# -*- coding: utf-8 -*-
# A层边界用例：每题3条（空输入/极值/异常），共15条。
import pytest
from p11_solutions import (Device, ex1_temp_convert, ex2_quadratic,
                           ex3_device_manager, ex4_plc_class, ex5_safe_reader)


# ---- 题1：温标换算的三条边界 ----
def test_ex1_absolute_zero():
    assert abs(ex1_temp_convert(-273.15)[1]) < 1e-9      # 绝对零度 -> 0K


def test_ex1_scales_cross_at_minus40():
    f, k = ex1_temp_convert(-40)
    assert f == -40.0 and abs(k - 233.15) < 1e-9         # 华氏/摄氏交点


def test_ex1_float_input_no_floor():
    assert abs(ex1_temp_convert(36.6)[0] - 97.88) < 1e-9  # 9/5 不能写成 9//5


# ---- 题2：根的三种判别式分支 ----
def test_ex2_no_real_root_returns_none():
    assert ex2_quadratic(1, 0, 1) is None


def test_ex2_double_root():
    assert ex2_quadratic(1, 2, 1) == (-1.0, -1.0)


def test_ex2_zero_a_is_not_quadratic():
    with pytest.raises(ZeroDivisionError):               # a=0 属非法输入
        ex2_quadratic(0, 2, 1)


# ---- 题3：字典增删改查的三条不变量 ----
def test_ex3_delete_effective():
    assert "DEV-01" not in ex3_device_manager()


def test_ex3_modify_effective():
    assert ex3_device_manager()["DEV-02"]["status"] == "运行"


def test_ex3_returns_new_dict_each_call():
    d1, d2 = ex3_device_manager(), ex3_device_manager()
    assert d1 == d2 and d1 is not d2                     # 无隐藏全局状态


# ---- 题4：继承/覆盖的三条 ----
def test_ex4_inherits_base():
    plc = ex4_plc_class()
    assert isinstance(plc, Device) and plc.status == "停止"


def test_ex4_start_overrides_and_mutates():
    plc = ex4_plc_class()
    assert plc.start() == "S7-1200启动, 扫描8ms" and plc.status == "运行"


def test_ex4_scan_ms_is_subclass_field():
    assert ex4_plc_class().scan_ms == 8
    assert not hasattr(Device("X", "y"), "scan_ms")        # 基类没有该字段


# ---- 题5：三条异常路径 ----
def test_ex5_missing_file(tmp_path):
    assert ex5_safe_reader(str(tmp_path / "nope.txt")) == "FILE_NOT_FOUND"


def test_ex5_bad_encoding(tmp_path):
    p = tmp_path / "bad.bin"
    p.write_bytes(b"\xff\xfe\x00abc")                     # 非法 utf-8 字节
    assert ex5_safe_reader(str(p)) == "BAD_ENCODING"


def test_ex5_reads_utf8(tmp_path):
    p = tmp_path / "ok.txt"
    p.write_text("产线温度 85.6℃", encoding="utf-8")
    assert ex5_safe_reader(str(p)) == "产线温度 85.6℃"
"""

td = tempfile.mkdtemp(prefix="w01_p11_")
with open(_os.path.join(td, "p11_solutions.py"), "w", encoding="utf-8") as fh:
    fh.write(_mod_src)
with open(_os.path.join(td, "test_p11_boundary.py"), "w", encoding="utf-8") as fh:
    fh.write(_test_src)

print("被测模块行数:", len(_mod_src.splitlines()), "｜用例数:",
      len(re.findall(r"^def test_", _test_src, re.M)), "（五题×3）")
r = subprocess.run([sys.executable, "-m", "pytest", "-q", "test_p11_boundary.py"],
                   cwd=td, capture_output=True, timeout=300)
print("pytest 目录:", td)
for line in (r.stdout or b"").decode("utf-8", "ignore").strip().splitlines()[-3:]:
    print("   ", line)
assert r.returncode == 0, "A层 pytest 全绿闸门未过\n" + (r.stdout or b"").decode(
    "utf-8", "ignore")[-1200:]
print("—— A层证据齐了：五题×3=15 条边界断言全部通过（被测代码 = 本 notebook 上方格子里的实现）——")
print("   待全班讲解选这条：test_ex4_scan_ms_is_subclass_field ——"
      " 子类新增字段不会污染基类，这是继承题最常见的写错方式。")
````

## 教师参考答案（来源：`教师参考/solutions.py` · `p11()`）

> **发放纪律：本附录勿随学生版下发。** 课堂按层放提示（D层演示照做/C层只给思路行/A层直接做）。

学生版 TODO 位置 → 下方参考实现的对应关系：

| 学生版 TODO 函数 | 教师完成态所在 cell | 关键实现要点 |
|---|---|---|
| ex1_temp_convert | 上方 Starter 2 | 套公式返回 `(F, K)`，`9/5` 为真除 |

> 下方 `p11()` 为五题参考答案（含 A 层扩展断言），与上方完成态 cell 互为校验；本 notebook 已逐 cell 运行通过。

````
class Device:                        # 题4基类(同学生版)
    def __init__(self, dev_id, name):
        self.dev_id, self.name = dev_id, name
        self.status = "停止"


def p11():
    """教师参考答案: 五题最小正确实现 + A层边界断言(可直接运行)"""
    import math

    def ex1(c):
        return c * 9 / 5 + 32, c + 273.15

    def ex2(a, b, c):
        d = b * b - 4 * a * c
        if d < 0:
            return None
        r = math.sqrt(d)
        return ((-b + r) / (2 * a), (-b - r) / (2 * a))

    def ex3():
        d = {}
        d["DEV-01"] = {"name": "空压机", "status": "运行"}
        d["DEV-02"] = {"name": "冷却泵", "status": "停止"}
        d["DEV-02"]["status"] = "运行"
        d.pop("DEV-01")
        return d

    class PLC(Device):
        def __init__(self, dev_id, name, scan_ms):
            super().__init__(dev_id, name)
            self.scan_ms = scan_ms

        def start(self):
            self.status = "运行"
            return f"{self.name}启动, 扫描{self.scan_ms}ms"

    def ex5(path):
        try:
            with open(path, encoding="utf-8") as f:
                return f.read()
        except FileNotFoundError:
            return "FILE_NOT_FOUND"
        except UnicodeDecodeError:
            return "BAD_ENCODING"

    assert ex1(100) == (212.0, 373.15)
    assert ex2(1, -3, 2) == (2.0, 1.0)
    assert ex3() == {"DEV-02": {"name": "冷却泵", "status": "运行"}}
    plc = PLC("DEV-01", "S7-1200", 8)
    assert isinstance(plc, Device) and plc.scan_ms == 8 and plc.status == "停止"
    assert plc.start() == "S7-1200启动, 扫描8ms" and plc.status == "运行"
    assert ex5("not_exist_file.txt") == "FILE_NOT_FOUND"
    assert ex1(-273.15) == (-459.67, 0.0) or abs(ex1(-273.15)[1]) < 1e-9
    assert ex2(1, 2, 1) == (-1.0, -1.0)
    print("P1.1 五题参考全部通过 ✓")
````
