### Starter 1：`# -*- coding: utf-8 -*-`
（Investigate: 读代码→Modify: 补全TODO）

````
# -*- coding: utf-8 -*-
"""P1.2 标准工程模板生成器 —— 运行后生成完整项目骨架
用法: python W02_template.py [项目名]
生成后: cd 项目名 && git init && pytest
"""
import os
import sys

FILES = {
    "src/main.py": '"""主程序入口"""\n\n\ndef main():\n    pass\n\n\nif __name__ == "__main__":\n    main()\n',
    "src/utils.py": '"""工具函数"""\n',
    "tests/test_main.py": '"""单元测试: pytest自动发现test_开头函数"""\nfrom src.main import main\n\n\ndef test_main_runs():\n    assert main() is None\n',
    "config/settings.yaml": "db_path: data/warehouse.db\npoll_interval: 0.5\n",
    "README.md": "# {name}\n\n## 快速开始\n```bash\npython -m venv venv && source venv/bin/activate\npip install -r requirements.txt\npytest\n```\n",
    "requirements.txt": "pandas>=1.5\nnumpy>=1.21\n",
    ".gitignore": "venv/\n__pycache__/\n*.pyc\n.DS_Store\ndata/*.db\n",
    ".pycharm/settings.json": '{"python.formatting.provider": "black",\n "python.linting.enabled": true}\n',
}
# === TODO-1(C层基线): 确认 FILES 的键都是"相对路径", 且 .gitignore 里已忽略 venv/ 与 __pycache__/ ===
# === TODO-2(B层·组长): 往 FILES 再补两份, 让生成的工程"依赖可还原 + 有工具链" ===
#     FILES["requirements-dev.txt"] = "pytest\nblack\nflake8\n"      # 开发依赖(与运行时依赖分开)
#     FILES["config/settings.yaml"] = "warn: 80\ndanger: 95\nseed: 42\n"  # 规则外置: 改配置不改代码
#   自检: 生成后运行 python -c "import ast; ast.parse(open('.../src/main.py').read())" 能过才算语法正确
# === TODO-3(A层·师傅): 再补 CI 与钩子两文件 ===
#     FILES[".github/workflows/ci.yml"]        = 安装依赖→pytest→flake8 三步的 Actions 工作流
#     FILES[".pre-commit-config.yaml"]         = commit 前自动 black + flake8
#   达标口径: Actions 绿钩 + pylint≥8.5(见工单)
````

### Starter 2：`def build(name):`
（Investigate: 读代码→Modify: 补全TODO）

````
def build(name):
    for rel, content in FILES.items():
        path = os.path.join(name, rel)
        os.makedirs(os.path.dirname(path), exist_ok=True)
        with open(path, "w", encoding="utf-8") as f:
            f.write(content.replace("{name}", name))
        print("创建", path)
    print(f"""
模板已生成 → {name}/
下一步(逐条执行, 记录到工单):
  1. cd {name} && git init
  2. git add . && git commit -m "feat: 初始化标准工程模板"
  3. git branch dev && git checkout dev        # 开发分支
  4. pip install pytest pylint && pytest       # 测试通过
  5. pylint src/ --fail-under=8.5              # 评分≥8.5(工单截图)
""")
    # === TODO-2 配套: 让 build() 自检——生成文件数应等于 len(FILES), 少一个就是路径写错 ===
    #   参考写法: made = [r for _, _, fs in os.walk(name) for r2 ...]; assert sorted(made) == sorted(FILES)
    #   教师版用 os.walk 复核真实磁盘结构, 并断言 requirements-dev.txt 里依赖带版本约束
````

### Starter 3：`if __name__ == "__main__":`
（Investigate: 读代码→Modify: 补全TODO）

````
if __name__ == "__main__":
    # notebook 里 sys.argv[1] 是本 notebook 的文件路径, 不能当项目名: 先过滤非法值
    raw = sys.argv[1] if len(sys.argv) > 1 else ""
    PROJECT = raw if raw.isidentifier() else "industrial_project"
    build(PROJECT)
    print("项目名:", PROJECT, "—— 改成 学号_姓名 再跑一次, 然后 cd 进去 git init")
````
