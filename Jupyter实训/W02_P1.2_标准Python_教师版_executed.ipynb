### Starter 1：`# -*- coding: utf-8 -*-`
（教师版：下方为 TODO 完成态，可直接运行；学生版此处留空待填）

````
# -*- coding: utf-8 -*-
"""P1.2 标准工程模板生成器 —— 教师完成态(含 A层pre-commit+CI / B层依赖可还原)
用法: python W02_template.py [项目名]
生成后: cd 项目名 && git init && pytest
"""
import os
import sys

MAIN_PY = (
    '"""主程序入口"""\n'
    '\n'
    '\n'
    'def main():\n'
    '    return 0\n'
    '\n'
    '\n'
    'if __name__ == "__main__":\n'
    '    raise SystemExit(main())\n'
)
UTILS_PY = (
    '"""工具函数"""\n'
    '\n'
    '\n'
    'def celsius_to_fahrenheit(c: float) -> float:\n'
    '    return c * 9 / 5 + 32\n'
)
TEST_MAIN = (
    '"""单元测试: pytest自动发现test_开头函数"""\n'
    'from src.main import main\n'
    'from src.utils import celsius_to_fahrenheit\n'
    '\n'
    '\n'
    'def test_main_runs():\n'
    '    assert main() == 0\n'
    '\n'
    '\n'
    'def test_convert():\n'
    '    assert celsius_to_fahrenheit(100) == 212\n'
)
README = (
    "# {name}\n"
    "\n"
    "## 快速开始\n"
    "```bash\n"
    "python -m venv venv && source venv/bin/activate\n"
    "pip install -r requirements.txt\n"
    "pytest\n"
    "```\n"
)
DEVELOP = (
    "# B层: 依赖可还原(运行/开发两套)\n"
    "-r requirements.txt\n"
    "pytest>=7.0\n"
    "pylint>=2.17\n"
    "black>=23.0\n"
    "flake8>=6.0\n"
    "pre-commit>=3.0\n"
)
PRECOMMIT = (
    "# A层: commit 前自动 black + flake8\n"
    "repos:\n"
    "  - repo: local\n"
    "    hooks:\n"
    "      - id: black\n"
    "        name: black\n"
    "        entry: black .\n"
    "        language: system\n"
    "        types: [python]\n"
    "      - id: flake8\n"
    "        name: flake8\n"
    "        entry: flake8 src tests --max-line-length=100\n"
    "        language: system\n"
    "        types: [python]\n"
)

CI_YML = (
    "# A层: push/PR 到 GitHub 后自动跑 装依赖→pytest→flake8（Actions 绿钩＝超优档证据）\n"
    "name: ci\n"
    "on: [push, pull_request]\n"
    "jobs:\n"
    "  test:\n"
    "    runs-on: ubuntu-latest\n"
    "    steps:\n"
    "      - uses: actions/checkout@v4\n"
    "      - uses: actions/setup-python@v5\n"
    "        with:\n"
    "          python-version: '3.11'\n"
    "      - run: pip install -r requirements.txt -r requirements-dev.txt\n"
    "      - run: pytest -q\n"
    "      - run: flake8 src tests --max-line-length=100\n"
)

FILES = {
    "src/__init__.py": "",
    "src/main.py": MAIN_PY,
    "src/utils.py": UTILS_PY,
    "tests/__init__.py": "",
    "tests/test_main.py": TEST_MAIN,
    "config/settings.yaml": "db_path: data/warehouse.db\npoll_interval: 0.5\n",
    "README.md": README,
    "requirements.txt": "pandas>=1.5\nnumpy>=1.21\nPyYAML>=6.0\n",
    "requirements-dev.txt": DEVELOP,
    ".pre-commit-config.yaml": PRECOMMIT,
    ".github/workflows/ci.yml": CI_YML,
    ".gitignore": "venv/\n__pycache__/\n*.pyc\n.DS_Store\ndata/*.db\n.pytest_cache/\n",
    ".pycharm/settings.json": '{"python.formatting.provider": "black",\n'
                              ' "python.linting.enabled": true}\n',
}
EXPECTED = sorted(FILES)
````

### Starter 2：`def build(name):`
（教师版：下方为 TODO 完成态，可直接运行；学生版此处留空待填）

````
def build(name):
    """生成标准骨架并自校验: 目录齐全 + requirements 可解析(A/B层要点)"""
    for rel, content in FILES.items():
        path = os.path.join(name, rel)
        os.makedirs(os.path.dirname(path), exist_ok=True)
        with open(path, "w", encoding="utf-8") as f:
            f.write(content.replace("{name}", name))
        print("创建", path)
    made = sorted(os.path.relpath(os.path.join(r, fn), name).replace(os.sep, "/")
                  for r, _, fs in os.walk(name) for fn in fs
                  if not r.endswith("__pycache__"))
    assert made == EXPECTED, f"骨架不完整: 缺 {set(EXPECTED) - set(made)}"
    for req in ("requirements.txt", "requirements-dev.txt"):
        with open(os.path.join(name, req), encoding="utf-8") as f:
            lines = [x.strip() for x in f if x.strip() and not x.startswith(("#", "-r"))]
        assert all(">=" in x or "==" in x for x in lines), req + " 必须钉版本下限"
    print(f"模板已生成 → {name}/  共 {len(EXPECTED)} 个文件 (结构自校验通过)")
    print("""
下一步(逐条执行, 记录到工单):
  1. cd 项目名 && git init
  2. git add . && git commit -m "feat: 初始化标准工程模板"   # A层: pre-commit 钩子自动 black+flake8
  3. git branch dev && git checkout dev                       # 开发分支
  4. pip install -r requirements-dev.txt && pytest            # B层: 依赖可还原
  5. pylint src/ --fail-under=8.5                             # 评分≥8.5(工单截图)
""")
    return name
````

````
if __name__ == "__main__":
    # 课堂: python W02_template.py industrial_project   → 在项目目录生成骨架
    # notebook: 生成到临时目录并自校验后清理, 不污染课程包(命令行参数也不误判)
    import shutil
    import tempfile
    tmp = tempfile.mkdtemp(prefix="p12_")
    try:
        build(os.path.join(tmp, "industrial_project"))
        print("模板生成 + 结构/依赖/钩子 自校验通过 ✓")
    finally:
        shutil.rmtree(tmp, ignore_errors=True)
````

### 💡 分层提示（先自己试5分钟再看）
**思路提示**
> - 结构：`src/` 放包代码 + `tests/` 放 `test_*.py` + `config/` 放 yaml，`run.py` 在根目录做入口
> - `FILES` 是"相对路径→文件内容"字典，`build()` 遍历它：`os.makedirs(dirname, exist_ok=True)` 再写
> - C层：`python -m venv venv` + `pip install -r requirements.txt` + `git init` 三连，venv 必须进 `.gitignore`
> - B层：拆 `requirements.txt`(运行) 与 `requirements-dev.txt`(开发, 用 `-r` 引用前者)，每行钉版本下限
> - A层：加 `.pre-commit-config.yaml`，`pre-commit install` 后 commit 前自动跑 black+flake8
> - 分支：`git branch dev && git checkout dev`，一个功能一个 commit，前缀 `feat:`/`fix:`
