## 教师参考答案（来源：`教师参考/solutions.py` · `p11()`）

> **发放纪律：本附录勿随学生版下发。** 课堂按层放提示（D层演示照做/C层只给思路行/A层直接做）。

学生版 TODO 位置 → 下方参考实现的对应关系：

| 学生版 TODO 函数 | 参考实现说明 |
|---|---|

```
def p11():
    import math

    def ex1(c):
        return c * 9 / 5 + 32, c + 273.15

    def ex2(a, b, c):
        d = b * b - 4 * a * c
        if d < 0:
            return None
        r = math.sqrt(d)
        return ((-b + r) / (2 * a), (-b - r) / (2 * a))

    def ex3():
        d = {}
        d["DEV-01"] = {"name": "空压机", "status": "运行"}
        d["DEV-02"] = {"name": "冷却泵", "status": "停止"}
        d["DEV-02"]["status"] = "运行"
        d.pop("DEV-01")
        return d

    class PLC:
        def __init__(self, dev_id, name, scan_ms):
            self.dev_id, self.name, self.scan_ms = dev_id, name, scan_ms
            self.status = "停止"

        def start(self):
            self.status = "运行"
            print(f"{self.name}启动, 扫描{self.scan_ms}ms")

    def ex5(path):
        try:
            with open(path, encoding="utf-8") as f:
                return f.read()
        except FileNotFoundError:
            return "FILE_NOT_FOUND"
        except UnicodeDecodeError:
            return "BAD_ENCODING"

    assert ex1(100) == (212.0, 373.15)
    assert ex2(1, -3, 2) == (2.0, 1.0)
    assert ex3() == {"DEV-02": {"name": "冷却泵", "status": "运行"}}
    assert ex5("x.txt") == "FILE_NOT_FOUND"
    print("P1.1 五题参考全部通过 ✓")
```
